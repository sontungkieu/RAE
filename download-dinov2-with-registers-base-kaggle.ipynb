```python
from pathlib import Path

repo_id = "facebook/dinov2-with-registers-base"
local_root = Path("/kaggle/working/pretrained")
local_dir = local_root / "dinov2-with-registers-base"
local_root.mkdir(parents=True, exist_ok=True)

print("repo_id:", repo_id)
print("local_dir:", local_dir)
```

```python
from huggingface_hub import snapshot_download
from transformers import AutoConfig, AutoImageProcessor, Dinov2WithRegistersModel

snapshot_path = snapshot_download(
    repo_id=repo_id,
    local_dir=str(local_dir),
    local_dir_use_symlinks=False,
    resume_download=True,
)

print("snapshot_path:", snapshot_path)
print("downloaded files:")
for path in sorted(local_dir.rglob("*")):
    if path.is_file():
        print(" ", path.relative_to(local_dir))

config = AutoConfig.from_pretrained(local_dir, local_files_only=True)
processor = AutoImageProcessor.from_pretrained(local_dir, local_files_only=True)
model = Dinov2WithRegistersModel.from_pretrained(local_dir, local_files_only=True)

print("config class:", type(config).__name__)
print("processor class:", type(processor).__name__)
print("model class:", type(model).__name__)
print("hidden_size:", getattr(config, "hidden_size", None))
print("num_hidden_layers:", getattr(config, "num_hidden_layers", None))
print("num_register_tokens:", getattr(config, "num_register_tokens", None))
```

```python
print("Use these local paths in sitdh config:")
print()
print(f"encoder_config_path: '{local_dir.as_posix()}'")
print("encoder_params:")
print(f"  dinov2_path: '{local_dir.as_posix()}'")
print()
print("Optional size check:")
print(f"du -sh {local_dir.as_posix()}")
```
